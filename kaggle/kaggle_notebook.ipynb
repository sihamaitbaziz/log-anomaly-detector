```python
import os, shutil
from pathlib import Path

# Clone OR copy the uploaded dataset. Edit the slug if you uploaded the repo as data.
if Path('/kaggle/working/repo/main.py').exists():
    os.chdir('/kaggle/working/repo')
elif Path('/kaggle/input').exists():
    inputs = list(Path('/kaggle/input').glob('*'))
    print('inputs:', inputs)
    for item in inputs:
        if (item / 'main.py').exists():
            dst = Path('/kaggle/working/repo')
            if not dst.exists():
                shutil.copytree(item, dst)
            os.chdir(dst)
            break
print('cwd', os.getcwd())
```
